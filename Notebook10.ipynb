```python
highest = 0
for i in range(5):
    name = input("Enter student name: ")
    m1 = int(input("Enter marks 1: "))
    m2 = int(input("Enter marks 2: "))
    m3 = int(input("Enter marks 3: "))
    total = m1 + m2 + m3
    percentage = total / 3
```

```python
    if percentage >= 75:
        grade = "A"
    elif percentage >= 60:
        grade = "B"
    elif percentage >= 40:
        grade = "C"
    else:
        grade = "F"
print("Name:", name)
```

```python
print("Total:", total)
```

```python
print("Percentage:", percentage)
```

```python
print("Grade:", grade)
```

```python
if percentage > highest:
        highest = percentage
        topper = name
```

```python
print("Topper:", topper)
```

```python
print("Highest Percentage:", highest)
```
